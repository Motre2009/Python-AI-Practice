# Звіт із продажів
У цьому розділі ми рахуємо підсумки за простими даними

| Variable | Value |
|---|---|
| Sum | 965 |
| Min value | 90 |
| Max value | 200 |
| Average value | 2.22 |

```python
sales = [120, 135, 100, 150, 90, 200, 170]
```

```python
sum_of_sales = sum(sales)
```

```python
min_value = min(sales)
```

```python
max_value = max(sales)
```

```python
average = max_value / min_value
```

```python
print('Sum:', sum_of_sales)
print('Min value:', min_value)
print('Max value:', max_value)
print('Average value:', average)
```
