## Оцінки студентів
Студенти та їх суперські бали

| Indicator | Value |
|---|---|
| Average score | 71.8 |
| Max score | 92 |
| Min score | 48 |

```python
students = [("Alex", 85), ("Maria", 92), ("John", 74), ("Olga", 60), ("Max", 48)]
```

```python
scores = []
weak_students = []
for name, score in students:
    scores.append(score)
    if (score <= 60):
        weak_students.append((name, score))

average = sum(scores) / len(scores)
max_score = max(scores)
min_score = min(scores)

print('Average score:', average)
print('Max score:', max_score)
print('Min score:', min_score)
print('Weak students:', weak_students)
```

```python
for name, score in students:
    stars_count = score // 10
    print(f"{name}| {'*' * stars_count}")
```
