## Practice:

**У вас є дані про книги з полями: "_Назва_", "_Автор_", "_Рік видання_", "_Ціна_".**

-   Створіть DataFrame з не менш ніж шістьма рядками.
-   Виведіть увесь DataFrame.
-   Визначте середню ціну книг.
-   Знайдіть книги, видані після 2015 року.
-   Відсортуйте DataFrame за ціною в порядку зростання.


```python
import pandas as pd

books = {
    "Назва": ["Кобзар", "1984", "Майстер і Маргарита", "Тіні забутих предків", "Місто", "Четверте крило"],
    "Автор": ["Тарас Шевченко", "Джордж Орвелл", "Михайло Булгаков", "Михайло Коцюбинський", "Валер'ян Підмогильний", "Ребекка Яррос"],
    "Рік видання": [1840, 1949, 1967, 1911, 1928, 2023],
    "Ціна": [250, 320, 410, 180, 290, 490]
}

df = pd.DataFrame(books)
df
```

```python
print(f'Середня ціна книг: {df["Ціна"].mean():.2f}')
```

```python
print(f'Книги виданні після 2015 року: \n{df[df["Рік видання"] > 2015]}')
```

```python
print(f'Відсортованна таблиця за ціною в порядку зростання: \n{df.sort_values("Ціна", ascending=True)}')
```
