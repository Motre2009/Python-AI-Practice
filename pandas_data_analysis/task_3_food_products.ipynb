## Practice:

**Є таблиця з даними про продукти харчування: "_Продукт_", "_Категорія_", "_Калорії_", "_Білки_".**

-   Створіть DataFrame з не менш ніж десятьма рядками.
-   Виведіть увесь DataFrame.
-   Знайдіть усі продукти з калорійністю вище 300.
-   Підрахуйте середню кількість білків за категоріями.
-   Відсортуйте DataFrame за калорійністю в порядку убування.


```python
import pandas as pd

food = {
    "Продукт": ["Куряче філе", "Рис", "Вівсянка", "Яйце", "Лосось", "Сир", "Банан", "Мигдаль", "Броколі", "Картопля", "Хліб", "Шоколад"],
    "Категорія": ["М'ясо", "Крупи", "Крупи", "Яйця", "Риба", "Молочне", "Фрукти", "Горіхи", "Овочі", "Овочі", "Випічка", "Солодощі"],
    "Калорії": [165, 130, 389, 155, 208, 402, 89, 579, 34, 77, 265, 546],
    "Білки": [31.0, 2.7, 16.9, 13.0, 20.0, 25.0, 1.1, 21.2, 2.8, 2.0, 9.0, 4.9]
}

df = pd.DataFrame(food)
df
```

```python
print(f'Усі продукти з калорійністю вище 300: \n{df[df["Калорії"] > 300]}')
```

```python
print(f'Середня кількість білків за категоріями: \n{df.groupby("Категорія")["Білки"].mean().round(2)}')
```

```python
print(f'Відсортованна таблиця за калорійністю в порядку убування: \n{df.sort_values("Калорії", ascending=False)}')
```
