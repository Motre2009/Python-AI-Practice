## Practice:

**Зібрано дані про співробітників та їхні проекти: "_Ім'я_", "_Проект_", "_Години_".**

-   Створіть DataFrame з не менш ніж вісьмома рядками.
-   Виведіть вихідний DataFrame.
-   Підрахуйте загальну кількість годин за кожним співробітником.
-   Підрахуйте загальну кількість годин за кожним проектом.
-   Знайдіть співробітника, який витратив найбільше годин.


```python
import pandas as pd

work = {
    "Ім'я": ["Олена", "Олена", "Іван", "Іван", "Петро", "Петро", "Марія", "Марія"],
    "Проект": ["Альфа", "Бета", "Альфа", "Гамма", "Бета", "Гамма", "Альфа", "Гамма"],
    "Години": [12, 8, 16, 10, 20, 5, 7, 14]
}

df = pd.DataFrame(work)
df
```

```python
print(f'Загальна кількість годин за кожним співробітником: \n{df.groupby("Ім'я")["Години"].sum()}')
```

```python
print(f'Загальна кількість годин за кожним проектом: \n{df.groupby("Проект")["Години"].sum()}')
```

```python
hours = df.groupby("Ім'я")["Години"].sum()
print(f'Співробітник, який витратив найбільше годин: \n{hours.idxmax()}: {hours.max()}')
```
