## Practice:

**У вас є таблиця продажів квитків: "_Фільм_", "_Місто_", "_Продані квитки_".**

-   Створіть DataFrame з не менш ніж дванадцятьма рядками.
-   Виведіть увесь DataFrame.
-   Підрахуйте загальну кількість проданих квитків за кожним фільмом.
-   Підрахуйте загальну кількість квитків по кожному місту.
-   Знайдіть фільм із найбільшою кількістю продажів.


```python
import pandas as pd

tickets = {
    "Фільм": ["Дюна", "Шоу Трумена", "Оппенгеймер", "Інтерстеллар", "Барбі", "Шоу Трумена", "Оппенгеймер", "Престиж", "Джуманджі: Поклик джунглів", "Втеча з Шоушенка", "На межі майбутнього", "Славні хлопці"],
    "Місто": ["Київ", "Львів", "Київ", "Одеса", "Київ", "Львів", "Одеса", "Львів", "Одеса", "Харків", "Харків", "Київ"],
    "Продані квитки": [120, 90, 150, 80, 200, 110, 60, 95, 75, 140, 130, 160]
}

df = pd.DataFrame(tickets)
df
```

```python
print(f'Загальна кількість проданих квитків за кожним фільмом: \n{df.groupby("Фільм")["Продані квитки"].sum()}')
```

```python
print(f'Загальна кількість проданих квитків по кожному місту: \n{df.groupby("Місто")["Продані квитки"].sum()}')
```

```python
tickets_by_movie = df.groupby("Фільм")["Продані квитки"].sum()
print(f'Фільм із найбільшою кількістю продажів: \n{tickets_by_movie.idxmax()}: {tickets_by_movie.max()}')
```
