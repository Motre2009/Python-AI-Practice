**Завдання:**
Створіть одновимірний масив із 10 випадкових чисел від 0 до 100.
-   Збережіть масив у змінну і виведіть його на екран.
-   Знайдіть мінімальне, максимальне та середнє значення масиву за допомогою вбудованих функцій NumPy.
-   Відсортуйте масив за зростанням і виведіть результат.
-   Відобразіть масив до і після сортування.

```python
import numpy as np
rng = np.random.default_rng()
```

```python
random_numbers = rng.integers(0, 100, size=10)
print(random_numbers)
```

```python
print(f'Min number: {random_numbers.min()}')
print(f'Max number: {random_numbers.max()}')
print(f'Average number: {random_numbers.mean()}')
```

```python
print(f'Array of random numbers before sort: {random_numbers}')
```

```python
sorted_array = np.sort(random_numbers)
print(f'Array of random numbers after sort: {sorted_array}')
```
