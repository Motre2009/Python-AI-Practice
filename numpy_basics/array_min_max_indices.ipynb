**Завдання:**
Створіть масив розміром 10×10, заповнений випадковими числами з плаваючою крапкою від 0 до 1.
-   Знайдіть мінімальне та максимальне значення в усьому масиві.
-   Визначте індекси мінімального та максимального елементів
    (наприклад: (рядок, стовпець)).
-   Виведіть сам масив і окремо знайдені значення разом з їхніми індексами.

```python
import numpy as np
rng = np.random.default_rng()
array_floats = rng.random(100).reshape(10, 10).round(3)
print(f"Array with random floating numbers: {array_floats}")
```

```python
print(f'Min value: {array_floats.min()}')
print(f'Index of min elem (rows, cols): {np.unravel_index(array_floats.argmin(), array_floats.shape)}')
print(f'Max value: {array_floats.max()}')
print(f'Index of max elem (rows, cols): {np.unravel_index(array_floats.argmax(), array_floats.shape)}')
```
