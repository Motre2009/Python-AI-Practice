**Завдання:**
Створіть два одновимірні масиви довжиною 5: перший заповніть числами від 1 до 5, другий — числами від 6 до 10.

-   Об'єднайте їх в один масив за допомогою np.concatenate.
-   Побудуйте двовимірний масив 2×5, у якому перший рядок — перший масив, а другий рядок — другий масив.
-   Об'єднайте обидва масиви в одну колонку (розмірність 5×2).
-   Виведіть усі три результати на екран.

```python
import numpy as np
first_array = np.arange(1, 6)
second_array = np.arange(6, 11)
full_array = np.concatenate((first_array, second_array))
full_array_in_rows = np.concatenate((first_array, second_array)).reshape(2, 5)
full_array_in_columns = np.column_stack((first_array, second_array))
```

```python
print(f'Full array: {full_array}')
print(f'Full array in rows: {full_array_in_rows}')
print(f'Full array in columns: {full_array_in_columns}')
```
