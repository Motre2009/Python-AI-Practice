```python
import numpy as np
```

```python
rows = int(input('Enter number of rows: '))
columns = int(input('Enter number of columns: '))
total = rows * columns

matrix = np.arange(total).reshape(rows, columns)
print(f'Original matrix: \n{matrix}')
```

```python
while True:
    new_rows = int(input('Enter rows of the new range of numbers: '))
    new_columns = int(input('Enter columns of the new range of numbers: '))
    count = new_rows * new_columns
    if count == total:
        break
    else:
        print(f'The new matrix contains {count} elements, but {total} are required')
    

new_matrix = matrix.reshape(new_rows, new_columns)
print(f'New matrix: \n{new_matrix}')
print(f'Row minimums: \n{new_matrix.min(axis=1)}')
print(f'Row maximums: \n{new_matrix.max(axis=1)}')
print(f'Sum of all elements: \n{new_matrix.sum()}')
```
