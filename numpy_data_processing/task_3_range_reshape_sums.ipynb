```python
import numpy as np
```

```python
while True:
    lower_limit = int(input('Enter lower limit of the range of numbers: '))
    upper_limit = int(input('Enter upper limit of the range of numbers: '))
    count = upper_limit - lower_limit + 1
    if count == 30:
        break
    else:
        print(f'The range contains {count} numbers, but 30 are required')
    
array = np.arange(lower_limit, upper_limit + 1)
print(f'Array created by user: \n{array}')
```

```python
matrix = array.reshape(6, 5)
print(f'Matrix 6x5: \n{matrix}')
print(f'The sum of array by rows: \n{matrix.sum(axis=1)}')
print(f'The max values in column: \n{matrix.max(axis=0)}')
```
