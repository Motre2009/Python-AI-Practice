```python
import numpy as np
```

```python
rng = np.random.default_rng()

lower_limit = int(input('Enter lower limit of the range of numbers: '))
upper_limit = int(input('Enter upper limit of the range of numbers: '))

array = rng.integers(lower_limit, upper_limit + 1, size=15)
print(f'Array: \n{array}')
print(f'Negative elements: \n{array[array < 0]}')
```

```python
copy_of_array = array.copy()
copy_of_array[copy_of_array < 0] = 0
print(f'Array with negatives replaced by zeros: {copy_of_array}')
print(f'Number of zeros in the modified array: {np.sum(copy_of_array == 0)}')
```
