```python
import numpy as np
```

```python
rng = np.random.default_rng()
array_in_range = rng.integers(0, 51, size=20)
print(f'An array in the range from 0 to 50: \n{array_in_range}')
```

```python
users_number = int(input('Enter number(threshold value): '))
print(f'Amount of array elements exceeding the value entered by the user: {np.sum(array_in_range > users_number)}')
print(f'Array elements exceeding the value entered by the user: {array_in_range[array_in_range > users_number]}')
```

```python
print(f'Max value in array: {array_in_range.max()}')
print(f'The first appearance of max element: {array_in_range.argmax()}')
```

```python
print(f'Sorted descending array: {np.sort(array_in_range, descending=True)}')
```
