```python
import numpy as np
```

```python
rng = np.random.default_rng()

lower_limit = int(input('Enter lower limit of the range of numbers: '))
upper_limit = int(input('Enter upper limit of the range of numbers: '))

matrix = rng.integers(lower_limit, upper_limit + 1, size=(5, 5))
print(f'Matrix: \n{matrix}')
```

```python
print(f'The main diagonal: \n{np.diagonal(matrix)}')
print(f'Sum of the main diagonal: \n{np.trace(matrix)}')
print(f'The matrix with zeros above the main diagonal: \n{np.tril(matrix)}')
```
