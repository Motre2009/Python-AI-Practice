    -   перший заповніть випадковими числами від 0 до 10;
    -   другий заповніть випадковими числами від 10 до 20.
-   Виведіть обидва вихідні масиви.
-   Об'єднайте масиви в один і виведіть результат.
-   Виконайте поелементне додавання та поелементну різницю вихідних масивів і виведіть обидва результати.

```python
import numpy as np
```

```python
rng = np.random.default_rng()

length_of_array = int(input('Enter length of array: '))
first_array = rng.integers(0, 10 + 1, size=length_of_array)
print(f'First array: \n{first_array}')

second_array = rng.integers(10, 20 + 1, size=length_of_array)
print(f'Second array: \n{second_array}')
```

```python
full_array = np.concatenate((first_array, second_array))
print(f'The full array: \n{full_array}')
print(f'Element-by-element addition: \n{first_array + second_array}')
print(f'Element-by-element difference: \n{first_array - second_array}')
```
