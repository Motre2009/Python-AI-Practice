Є список студентів із зазначенням: "_Ім'я_", "_Курс_", "_Оцінка_".

-   Створіть DataFrame і заповніть його не менш ніж вісьмома рядками.
-   Виведіть вихідний DataFrame.
-   Обчисліть середню оцінку за кожним курсом.
-   Знайдіть максимальну та мінімальну оцінку за кожним курсом.

```python
import pandas as pd
```

```python
data = {
    "Ім'я": ["Олег", "Марина", "Богдан", "Іванна", "Максим",
             "Софія", "Артем", "Христина", "Роман", "Вікторія"],
    "Курс": [1, 2, 1, 3, 2, 1, 3, 2, 1, 3],
    "Оцінка": [85, 92, 78, 88, 95, 67, 90, 73, 81, 99]
}

df = pd.DataFrame(data)
df
```

```python
print(f"Середня оцінка за кожним курсом:\n{df.groupby('Курс')['Оцінка'].mean()}")
```

```python
print(f"Максимальна оцінка за кожним курсом:\n{df.groupby('Курс')['Оцінка'].max()}")
print(f"Мінімальна оцінка за кожним курсом:\n{df.groupby('Курс')['Оцінка'].min()}")
```
