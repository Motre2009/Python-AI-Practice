```python
import pandas as pd
```

```python
data = {
    "Товар": ["Ноутбук", "Мишка", "Клавіатура", "Монітор", "Навушники",
              "Принтер", "Веб-камера", "Флешка", "Роутер", "Планшет"],
    "Категорія": ["Електроніка", "Аксесуари", "Аксесуари", "Електроніка", "Аксесуари",
                  "Електроніка", "Аксесуари", "Аксесуари", "Електроніка", "Електроніка"],
    "Ціна": [25000, 350, 800, 7500, 1200, 4500, 600, 250, 1800, 12000],
    "Кількість": [3, 15, 10, 5, 8, 2, 6, 20, 4, 3]
}

df = pd.DataFrame(data)
df
```

```python
df['Виручка'] = df['Ціна'].multiply(df['Кількість'])
print(f"Oновлений DataFrame:\n{df}")
```
