```python
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange", "watermelon"]
```

```python
votes_count = {}
```

```python
for fruit in votes:
    if fruit in votes_count:
        votes_count[fruit] += 1
    else:
        votes_count[fruit] = 1

print(votes_count)
```

```python
leader = ''
max_count = 0
for fruit, count in votes_count.items():
    print(f"{fruit}: {count}")
    
    if count > max_count:
        max_count = count
        leader = fruit

print(f"Leader: {leader}")
```

```python
def get_count(pair):
    return pair[1]
sorted_vote = sorted(votes_count.items(), key=get_count, reverse=True)
print(sorted_vote[:3])
```
