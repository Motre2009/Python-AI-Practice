## Results:
| Indicator | Value |
|---|---|
| Sum of expenses | 6400 |
| Average value | 914.29 |
| Max value of expenses | 1500 |
| Min value of expenses | 400 |
| Number of days when consumption was higher than 1000 | 2 |

```python
expenses = [1200, 1500, 800, 600, 1000, 400, 900]
sumOfExpenses = sum(expenses)
average = sum(expenses) / len(expenses)
maxValueOfExpenses = max(expenses)
minValueOfExpenses = min(expenses)

countOfDays = 0
for x in expenses:
    if (x > 1000):
        countOfDays = countOfDays + 1

print('Sum of expenses:', sumOfExpenses)
print('Average value', round(average, 2))
print('Max value of expenses', maxValueOfExpenses)
print('Min value of expenses', minValueOfExpenses)
print('Number of days when consumption was higher than 1000:', countOfDays)
```
