## Results:
| Indicator | Value |
|--- |---|
| Average value | 4.0 |
| Max value | 7 |
| Min value | 1 |
| Difference of max value and min value | 6 |

```python
temperatures = [2, 5, 4, 6, 7, 3, 1]
average = sum(temperatures) / len(temperatures)
max_temp = max(temperatures)
min_temp = min(temperatures)
difference = max_temp - min_temp
print(temperatures)
print('Average value:', average)
print('Max value:', max_temp)
print('Min value:', min_temp)
print('Difference of max value and min value:', difference)
```
