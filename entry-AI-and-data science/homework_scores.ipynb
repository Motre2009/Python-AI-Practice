### Better lesson: English

```python
math = [80, 92, 75, 60, 88]    
physics = [70, 65, 90, 85, 78]    
english = [95, 100, 85, 90, 87]
```

```python
average_grade_of_math = sum(math) / len(math)
average_grade_of_physics = sum(physics) / len(physics)
average_grade_of_english = sum(english) / len(english)
```

```python
max_grade_of_math = max(math)
max_grade_of_physics = max(physics)
max_grade_of_english = max(english)
```

```python
min_grade_of_math = min(math)
min_grade_of_physics = min(physics)
min_grade_of_english = min(english)
```

```python
average_grades_of_lessons = {
    "Math": average_grade_of_math,
    "Physics": average_grade_of_physics,
    "English": average_grade_of_english
}
```

```python
better_lesson = ""
max_grade = 0

for lesson, grade in average_grades_of_lessons.items():
    if grade > max_grade:
        max_grade = grade
        better_lesson = lesson
```

```python
print(f"Average grade of math: {average_grade_of_math}")
print(f"Average grade of physics: {average_grade_of_physics}")
print(f"Average grade of english: {average_grade_of_english}")
```

```python
print(f"Max grade of math: {max_grade_of_math}")
print(f"Max grade of physics: {max_grade_of_physics}")
print(f"Max grade of english: {max_grade_of_english}")
```

```python
print(f"Min grade of math: {min_grade_of_math}")
print(f"Min grade of physics: {min_grade_of_physics}")
print(f"Min grade of english: {min_grade_of_english}")
```

```python
print(f"Better lesson: {better_lesson}")
```
