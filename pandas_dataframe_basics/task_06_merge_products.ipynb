Зібрано дані про товари та продажі у двох таблицях. Перша містить: "_ID_", "_Назва_", "_Ціна_". Друга містить: "_ID_", "_Кількість_".

-   Створіть два DataFrame і заповніть їх не менш ніж п'ятьма рядками.
-   Виведіть обидва DataFrame.
-   Об'єднайте їх за колонкою "_ID_".
-   Додайте новий стовпець "_Сума_" як добуток ціни та кількості.
-   Виведіть підсумковий DataFrame.

```python
import pandas as pd

products = {
    "ID": [1, 2, 3, 4, 5],
    "Назва": ["Ноутбук", "Мишка", "Клавіатура", "Монітор", "Телефон"],
    "Ціна": [25000, 800, 1500, 7000, 18000]
}

sales = {
    "ID": [1, 2, 3, 4, 5],
    "Кількість": [5, 30, 20, 8, 12]
}

df_products = pd.DataFrame(products)
df_products
```

```python
df_sales = pd.DataFrame(sales)
df_sales
```

```python
df = df_products.merge(df_sales, on='ID', how='inner')
df
```

```python
df['Сума'] = df['Ціна'].multiply(df['Кількість'])
df
```
