Доступні дані про продажі з колонками: "_Товар_", "_Категорія_", "_Ціна_", "_Кількість_".

-   Створіть DataFrame з не менш ніж десятьма рядками.
-   Виведіть вихідний DataFrame.
-   Додайте новий стовпець "_Виручка_" як добуток ціни та кількості.
-   Виведіть оновлений DataFrame.

```python
import pandas as pd

sales = {
    "Товар": ["Ноутбук", "Мишка", "Клавіатура", "Монітор", "Телефон", "Планшет", "Навушники", "Колонка", "Принтер", "Веб-камера"],
    "Категорія": ["Електроніка", "Аксесуари", "Аксесуари", "Електроніка", "Електроніка", "Електроніка", "Аксесуари", "Аудіо", "Офіс", "Аксесуари"],
    "Ціна": [25000, 800, 1500, 7000, 18000, 12000, 2500, 3200, 6000, 1100],
    "Кількість": [5, 30, 20, 8, 12, 10, 25, 15, 4, 22]
}

df = pd.DataFrame(sales)
df
```

```python
df['Виручка'] = df['Ціна'].multiply(df['Кількість'])
print(f"Oновлений DataFrame:\n{df}")
```
