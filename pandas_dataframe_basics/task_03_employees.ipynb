Є таблиця співробітників із колонками: "_Ім'я_", "_Вік_", "_Відділ_", "_Зарплата_".

-   Створіть DataFrame з не менш ніж п'ятьма рядками.
-   Виведіть усі рядки, де зарплата вища за 50000.
-   Відсортуйте таблицю за віком у порядку зростання.
-   Відсортуйте таблицю за зарплатою в порядку спадання.

```python
import pandas as pd

employees = {
    "Ім'я": ["Іван", "Олена", "Петро", "Марія", "Юлія", "Андрій"],
    "Вік": [25, 32, 41, 28, 36, 45],
    "Відділ": ["IT", "HR", "IT", "Фінанси", "Маркетинг", "IT"],
    "Зарплата": [60000, 45000, 72000, 51000, 48000, 85000]
}

df = pd.DataFrame(employees)
df
```

```python
print(f"Працівники, у яких зарплата вища за 50000: \n{df[df['Зарплата'] > 50000]}")
```

```python
print(f"Таблиця за віком у порядку зростання: \n{df.sort_values('Вік', ascending=True)}")
```

```python
print(f"Таблиця за зарплатою в порядку спадання: \n{df.sort_values('Зарплата', ascending=False)}")
```
